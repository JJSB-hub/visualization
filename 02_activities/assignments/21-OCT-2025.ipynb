```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y)
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y)
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.barh(x,y)
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y)
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

font1 = {'family' : 'sans-serif', 'color':'blue', 'size':20}
font2 = {'family' : 'monospace', 'color':'green', 'size':14}

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('Total growth over time',fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Year since start', fontdict = font2)

```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y, 
           marker = '*', 
           color = 'indigo')
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
           marker = '*', 
           color = 'indigo',
           linestyle = '--',
           linewidth = 2)
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
           marker = '*', 
           markersize = 12,
           color = '#7425b9',
           linestyle = '--',
           linewidth = 2,
           markeredgecolor = '#fa9359',
           markerfacecolor = "#000000")
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
           marker = '*', 
           markersize = 12,
           color = '#7425b9',
           linestyle = '--',
           linewidth = 2,
           markeredgecolor = '#fa9359',
           markerfacecolor = "#000000")

ax.grid(axis = 'y', color = 'chartreuse', linestyle = '--', linewidth = 1)
```
