```python
#Set variables
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y2 = np.array([110,180,240,99,220])

#Build figure container
fig, (ax1, ax2) = plt.subplots(ncols = 2,
                               nrows = 1,
                               figsize=(7,3))

#Set up axes
ax1.scatter(x1,y1, label = 'Person 1')
#Add legend; set parameters
ax1.legend(loc = 'lower right')

#Set up axes
ax2.bar(x2,y2)
```

```python
# Create Mosaic layout
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7,4),
                                    layout = 'constrained')

#Place axes within mosaic
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

#Add labels to each axes
someaxes['ax1'].set_xlabel('A big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
#Set variables
x = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1 = np.array([110,180,240,99,220])
y2 = np.array([170,100,90,120,50])

#Build figure container with 2 axes
fig, ax = plt.subplots(figsize=(7,3))
ax.bar(x,y1, color = 'indigo')
ax.plot(x,y2,color = 'red')
```

```python
#Error variable
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2,color = 'red')
ax.errorbar(x,y2,
            yerr = y2_sd,
            fmt = 'none',
            ecolor = 'indigo',
            elinewidth = 4,
            capsize =6,
            capthick = 4,
            errorevery = 2)
```

```python
#import relevant sublibraries
from PIL import Image
import requests
from io import BytesIO
```

```python
#download image from the internet
response = requests.get('https://static.wikia.nocookie.net/the-ultimate-good/images/3/3d/Luffy.png/revision/latest?cb=20230417200546')
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
#Build figure container and set axes
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2, color = 'red')

#set image parameters/location
ax_image = fig.add_axes([0.1,
                         0.11,
                         0.15,
                         0.35])

#plot image
ax_image.imshow(image)
ax_image.axis('off')
```
